```python
import os

# Get the directory path from the 'dir' variable
directory_path = dir

# List all files in the directory
all_files = sorted([f for f in os.listdir(directory_path) if os.path.isfile(os.path.join(directory_path, f))])

print(f"Found {len(all_files)} files to rename.")

# Rename each file sequentially
for i, filename in enumerate(all_files):
    old_file_path = os.path.join(directory_path, filename)

    # Split the filename into name and extension
    name, ext = os.path.splitext(filename)

    # Convert .jpeg to .jpg if found
    if ext.lower() == '.jpeg':
        new_ext = '.jpg'
    else:
        new_ext = ext

    new_filename = f"Muntean_Larisa_Maria{i+1:04d}{new_ext}"
    new_file_path = os.path.join(directory_path, new_filename)

    try:
        os.rename(old_file_path, new_file_path)
        print(f"Renamed '{filename}' to '{new_filename}'")
    except OSError as e:
        print(f"Error renaming '{filename}': {e}")

print("File renaming complete.")
```

```python
import os
import shutil
import random

# Define paths
src_root = '/content/drive/MyDrive/Colab Notebooks/Licenta/lfw'
dest_root = '/content/drive/MyDrive/Colab Notebooks/Licenta/lfw50'

# Create destination root if it doesn't exist
if not os.path.exists(dest_root):
    os.makedirs(dest_root)
    print(f"Created directory: {dest_root}")

# Iterate through folders in lfw
subdirs = [d for d in os.listdir(src_root) if os.path.isdir(os.path.join(src_root, d))]

print(f"Found {len(subdirs)} total directories. Processing...")

processed_count = 0
for subdir in subdirs:
    src_subdir_path = os.path.join(src_root, subdir)
    dest_subdir_path = os.path.join(dest_root, subdir)

    # Get all files in the subdirectory
    files = [f for f in os.listdir(src_subdir_path) if os.path.isfile(os.path.join(src_subdir_path, f))]

    # Check if folder has at least 50 photos
    if len(files) >= 50:
        # Create the subfolder in lfw50
        os.makedirs(dest_subdir_path, exist_ok=True)

        # Randomly pick 50 photos
        selected_files = random.sample(files, 50)

        # Copy files to the new location
        for file_name in selected_files:
            shutil.copy2(os.path.join(src_subdir_path, file_name), os.path.join(dest_subdir_path, file_name))

        processed_count += 1
        if processed_count % 10 == 0:
            print(f"Processed {processed_count} folders...")

print(f"Finished! Created 'lfw50' with {processed_count} directories containing exactly 50 photos each.")
```

```python
import os

# Path to the processed dataset
dest_root = '/content/drive/MyDrive/Colab Notebooks/Licenta/lfw50'

# Get all directory names (labels) and sort them alphabetically
labels = sorted([d for d in os.listdir(dest_root) if os.path.isdir(os.path.join(dest_root, d))])

# Path for the output text file
labels_file_path = os.path.join(os.path.dirname(dest_root), 'labels.txt')

# Write labels to the file
with open(labels_file_path, 'w') as f:
    for label in labels:
        f.write(f"{label}\n")

print(f"Successfully saved {len(labels)} labels to: {labels_file_path}")
# Display the labels
print("Labels:", labels)
```
